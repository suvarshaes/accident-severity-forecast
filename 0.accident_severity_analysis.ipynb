```python
# Import required libraries
import pandas as pd
import numpy as np
```

## Data Loading 


```python
# Load the data (update the path to your data file)
# data = pd.read_csv('data/raw/US_Accidents_March23.csv')


# Define the size of your chunks
chunk_size = 100000 

# Create an empty list to hold processed chunks
processed_chunks = []

# Replace 'US_Accidents.csv' with the actual file path
for chunk in pd.read_csv('data/raw/US_Accidents_March23.csv', chunksize=chunk_size):
    # Perform your analysis or processing on each chunk
    # For example, filter data for a specific state
    filtered_chunk = chunk[chunk['State'] == 'NY']
    
    # Store the processed chunk
    processed_chunks.append(filtered_chunk)
    
# After the loop, concatenate the list of processed chunks into a single DataFrame
final_df = pd.concat(processed_chunks, ignore_index=True)

# Now you can work with the smaller, processed DataFrame
print(final_df.head())

#Display basic information about the dataset
# print("Dataset Info:")
# print(data.info())
# print("\nFirst few rows:")
# data.head()
```

```python
final_df.to_csv('data/raw/US_Accidents_March23_NY.csv', index=False)
```
